```python
#Problem :
side1=input("Enter the side length:")
side2=input("Enter the side length:")
side3=input("Enter the side length:")
side1=int(side1)
side2=int(side2)
side3=int(side3)

def check_rightangletriangle(side1,side2,side3):
    if (side1**2)==(side2**2)+(side3**2):
        print("The triangle is right angle")
    elif  (side2**2)==(side1**2)+(side3**2):
        print("The triangle is right angle")
    elif  (side3**2)==(side1**2)+(side2**2):
        print("The triangle is right angle")
    else:
        print("The triangle is not right angle")

check_rightangletriangle(side1,side2,side3)
```
