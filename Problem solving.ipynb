```python
"""
Even and Odd Numbers
Create a list of 10 numbers. Separate the numbers into two lists: one containing even numbers and the other containing odd numbers.
"""
numbers=[2,5,6,7,8,9,14,3,67,89]
evennumbers=[]
oddnumbers=[]
for i in range (0,10):
    if numbers[i]%2==0:
        evennumbers.append(numbers[i])
    else:
        oddnumbers.append(numbers[i])
print(evennumbers)
print(oddnumbers)
```

```python
"""
Tuple Operations
Consider the tuple:
numbers = (10, 20, 30, 20, 40, 20, 50)
Write a program to:
Count the occurrence of 20
Check whether 40 is present
Find the position of 30
"""
numbers = (10, 20, 30, 20, 40, 20, 50)
occurance=numbers.count(20)
print(occurance)
target_number = 30
if target_number in numbers:
     print(f"{target_number} is present in the tuple!")
else:
    print(f"{target_number} is NOT present in the tuple.")
position=numbers.index(30)
print("30 is present in the",(position+1),"position in tuple")
```

```python
"""
 Grade Calculator
Accept marks from the user and display the grade according to the following:
90–100 → A
75–89 → B
60–74 → C
40–59 → D
Below 40 → F
Display "Invalid marks" if the marks are outside the range 0–100.
"""

marks =int(input("Enter the marks")) 

print(f"Marks entered: {marks}")
if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
elif marks >= 40:
    print("Grade: D")
else:
    print("Grade: F")
```
