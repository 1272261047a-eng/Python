```python
#Problem 1:Print numbers 1 to 50 using for loop problem
for i in range(1,51):
    print(i)
```

```python
# Problem 2 :Print number 50 to 1 using while loop
i =50
while i>=1:
    print(i)
    i=i-1
```

```python
# Problem 3:Print multiples of 5 from 5 to 50
for i in range(5,51,5):
    print(i)
    
```

```python
#Problem 4:Print a word 5 times using while loop
word="Aditya"
i=0
while i<5:
    print(word)
    i+=1
```

```python
#Problem 5:Print each character of a word
word="Aditya"
for char in word:
    print(char)
```

```python
#Problem 6:Print numbers 10 to 1 using for loop
for i in range(10,0,-1):
    print(i)
```

```python
#Problem 7:Print even
for i in range(0,101,2):
    print(i)
```

```python
#Problem 8 Print odd
for i in range(1,100,2):
    print(i)
```

```python
#Problem 9:Print number from 1 to 10 ,stop at 6
for i in range(0,11):
    print(i)
    if i==6:
        break
```

```python
#Problem 10:Print number from 1 to 10 ,Skip 5
for i in range(0,11):
    if i ==5:
        continue
    print(i)
```

```python
#Two number calculator

def addition(x,y):
    print(x+y)

def subtraction(x,y):
    print(x-y)

def multiplication(x,y):
    print(x*y)

def division(x,y):
    print(x/y)
    
def modulus(x,y):
    print(x%y)

addition(5,4)
subtraction(5,4)
multiplication(5,4)
division(8,2)
modulus(3,2)
```
