```python
#list
studentlist = ["Aditya","Aayush","Jagganath","Dyvanshu"]
#print the items in the list
print(studentlist)
#list length
print(len(studentlist))
# type of the list
print(type(studentlist))
# we can define the list using list()
marks = list((97,78,81))
print(marks)
```

```python
# acess of items from list
studentlist = ["Aditya","Aayush","Jagganath","Dyvanshu"]
print(studentlist[0])
print(studentlist[-1])
print(studentlist[0:3])
print(studentlist[1:])
print(studentlist[:2])
print(studentlist[-4:-2])
```

```python
# check if item in the list exist
studentlist = ["Aditya","Aayush","Jagganath","Dyvanshu"]
if "Aditya" in studentlist:
  print("Yes ,'Aditya' is in student list")
else:
  print("No ,'Aditya' is not in student list")
```

```python
#change or update list items
studentlist = ["Aditya","Aayush","Jagganath","Dyvanshu"]
studentlist[1]="Vendant"
print(studentlist)
studentlist[2:4]=["Laxmi","Sammit"]
print(studentlist)
studentlist.insert(2,"Shuban")
print(studentlist)
#insert functions allow us to add item anywhere,append function adds at the end
studentlist.append("Rani")
print(studentlist)
```

```python
#use removs,pop,del to delete the list items
studentlist = ["Aditya","Aayush","Jagganath","Dyvanshu","Laxmi","Sammit"]
studentlist.remove("Jagganath")
print(studentlist)
studentlist.pop(1)
print(studentlist)
studentlist.pop()
print(studentlist)
del studentlist[2]
print(studentlist)
```

```python
#Tuple
studentlist = ("Aditya","Aayush","Jagganath","Dyvanshu")
print(studentlist)
print(type(studentlist))

#add elements to tuple
temp_list=list(studentlist)
temp_list.append("Sammit")
studentlist=tuple(temp_list)
print(studentlist)

#update elements in tuple
list1=list(studentlist)
list1[1]="Laxmi"
studentlist=tuple(list1)
print(studentlist)

#delete the elements in tuple
list2=list(studentlist)
list2.pop(2)
studentlist=tuple(list2)
print(studentlist)
'''
del studentlist
print(studentlist)
'''
#the code will show an error because the tuple does not exist 
```
