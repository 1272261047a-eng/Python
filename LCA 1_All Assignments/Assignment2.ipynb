```python
#code for largest of 3 numbers
x=int(input("Enter the value of x:"))
y=int(input("Enter the value of y:"))
z=int(input("Enter the value of z:"))
if x>y and x>z:
  print("x is the largest number")
elif y>x and y>z:
  print("y is the largest number")
elif z>x and z>y:
  print("z is the largest number")
elif x==y and y==z:
  print("x,y and z are equal")
elif x==y>z:
  print("x and y are equal and larger than z")
elif y==z>x:
  print("y and z are equal and larger than x")
elif x==z>y:
  print("x and z are equal and larger than y")
else:
  print("Enter a valid input")
```
