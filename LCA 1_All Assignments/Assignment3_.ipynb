```python
#Problem :To check if the triangle is right angle triangle 

side1=input("Enter the side length:")
side2=input("Enter the side length:")
side3=input("Enter the side length:")
side1=int(side1)
side2=int(side2)
side3=int(side3) 
 

def check_rightangletriangle(side1,side2,side3):
    sides=[side1,side2,side3]
    sides.sort()
    if sides[2]**2==sides[1]**2+sides[0]**2:
        print("The triangle is right angle")
    else:
        print("The triangle is not right angle triangle")

check_rightangletriangle(side1,side2,side3)
```
