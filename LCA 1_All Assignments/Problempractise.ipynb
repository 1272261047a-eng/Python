```python
#Problem 7: Marks Tuple

marks = (78, 90, 65, 88, 90, 72)
templist=list(marks)
templist.sort()
marks=tuple(templist)
print("Maximum no. ",marks[5])
print("Minimum no.",marks[0])
j = 0

for i in range (0,len(marks)):
    if marks[i]==90:
        j+=i
print("90 occurs",j,"times in tuple")
position=marks.index(88)
print("88 is present at:",(position+1))
print(marks[0:4])
```
