```python
import numpy as np
marks = [10,20,30,40,50]
marks2 = np.array(marks)
print(marks2)
```

```python
a=np.array([10,20,30,40,50])
b=np.array([20,30,40,50,60])
print(a+b)
print(b-a)
print(a*b)
print(b/a)
```

```python
#marks=[10,20,30,40,50]
#print(marks+5)
#we will get error from this type of operation
marks=np.array([10,20,30,40,50])
new_marks=marks+5
print(new_marks)
print(type(marks))
```

```python
#2D array
marks = np.array([[1,2,3],[8,3,0],[1,2,3]])
print(marks+2)
```

```python
marks = np.array([10,20,30,40,50])
print(np.sum(marks))
print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
print(np.argmax(marks))
print(np.argmin(marks))
print(np.sqrt(marks))
print(marks**2)
```

```python
marks = np.array([[1,2,0],[4,2,2],[1,5,3]])
print(np.sum(marks))
print(np.sum(marks,axis=0))#addition of column
print(np.sum(marks,axis=1))#addition of row 
```

```python
#slicing of array
marks = np.array([10,20,30,40,50])
print(marks[0:4])
print(marks[1:3])
print(marks[0:])
print(marks[:2])
```

```python
marks=np.array([10,20,30,40,50,60,70])
print(marks>50)
print(marks<50)
print(marks==30)
```

```python
marks=np.array([10,20,30,40,50,60,70,80])
print(marks[-3:])
print(marks[:-4])
print(marks[-4:-1])
print(marks[-1:-4])
#slicing with addition of step
print(marks[0:4:2])
print(marks[0:5:2])
print(marks[-6:-1:2])
```
